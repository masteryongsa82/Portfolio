```python
#AES algorithm

try:
    from Crypto.Cipher import AES
    from Crypto.Random import get_random_bytes
except:
    !pip install pycryptodome

from PIL import Image
import matplotlib.pyplot as plt


def aes_encrypt(text):
    key = get_random_bytes(32)
    cipher = AES.new(key, AES.MODE_GCM)

    ciphertext, tag = cipher.encrypt_and_digest(text.encode())


    return cipher.nonce + ciphertext + tag


def bytes_to_image(byte_data):
    W, H = 128, 128
    needed = W * H * 3


    padded = (byte_data * (needed // len(byte_data) + 1))[:needed]

    pixels = []
    for i in range(0, needed, 3):
        pixels.append((padded[i], padded[i+1], padded[i+2]))

    img = Image.new("RGB", (W, H))
    img.putdata(pixels)

    plt.figure(figsize=(4,4))
    plt.imshow(img)
    plt.axis("off")
    plt.title("AES Visualization")
    plt.show()


msg = input("메시지 입력: ")

aes_bytes = aes_encrypt(msg)

bytes_to_image(aes_bytes)
```

```python
#caeser cipher

try:
    from PIL import Image
except:
    !pip install pillow

import math
import matplotlib.pyplot as plt


def caesar_encrypt(text, shift=3):
    result = ""
    for ch in text:
        if ch.isalpha():
            base = 65 if ch.isupper() else 97
            result += chr((ord(ch) - base + shift) % 26 + base)
        else:
            result += ch
    return result


def text_to_image(text, filename="caesar.png"):
    byte_data = text.encode()

    size = math.ceil(math.sqrt(len(byte_data)//3 + 1))
    padded = byte_data + bytes([0]*(size*size*3 - len(byte_data)))

    pixels = []
    for i in range(0, len(padded), 3):
        pixels.append([padded[i], padded[i+1], padded[i+2]])

    img = Image.new("RGB", (size, size))
    img.putdata([tuple(p) for p in pixels])
    img.save(filename)


    plt.imshow(img)
    plt.axis("off")
    plt.show()

    print(f"{filename} 저장됨")


msg = input("메시지 입력: ")
shift = int(input("shift 값: "))

cipher = caesar_encrypt(msg, shift)
print("\n카이사르 암호:\n", cipher)

text_to_image(cipher)
```
